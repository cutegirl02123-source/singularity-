```python
import numpy as np
from scipy.optimize import least_squares, root
import jax
import jax.numpy as jnp
import warnings

warnings.filterwarnings('ignore', category=UserWarning)
jax.config.update("jax_enable_x64", True)
print("Environment initialized with JAX x64 precision enabled.")
```

```python
def system_vector_field_jax_4d(state, p):
    A, R, N, M = state
    alpha, beta, gamma, rho, R_max, eta, omega, kappa, S, K, mu, memory_gain, sigma, delta_m, eps_m = (
        p['alpha'], p['beta'], p['gamma'], p['rho'], p['R_max'], p['eta'],
        p['omega'], p['kappa'], p['S'], p['K'], p['mu'], p['memory_gain'], p['sigma'], p['delta_m'], p['eps_m']
    )

    bounded_memory_term = gamma * jnp.tanh(sigma * M)

    dA_dt = alpha * N - beta * R * A + bounded_memory_term
    dR_dt = rho * (R_max - R) - (eta * A) / (1.0 + omega * A)
    dN_dt = kappa * A * S * (1.0 - N / K) - mu * N
    dM_dt = memory_gain * (A - M) + delta_m * (M ** 3) - eps_m * (M ** 5)

    return jnp.array([dA_dt, dR_dt, dN_dt, dM_dt])

jac_state_jax_4d = jax.jit(jax.jacfwd(system_vector_field_jax_4d, argnums=0))
hess_state_jax_4d = jax.jit(jax.hessian(system_vector_field_jax_4d, argnums=0))
third_state_jax_4d = jax.jit(jax.jacfwd(jax.hessian(system_vector_field_jax_4d, argnums=0), argnums=0))

def directional_d_tensor(state, p, u, v, w, x):
    f = lambda s: system_vector_field_jax_4d(s, p)
    d1 = lambda s: jax.jvp(f, (s,), (u,))[1]
    d2 = lambda s: jax.jvp(d1, (s,), (v,))[1]
    d3 = lambda s: jax.jvp(d2, (s,), (w,))[1]
    _, d4_val = jax.jvp(d3, (state,), (x,))
    return d4_val
```

```python
p_base = {
    'alpha': 1.2, 'beta': 0.8, 'gamma': 1.0, 'rho': 0.6, 'R_max': 5.0,
    'eta': 0.5, 'omega': 0.1, 'kappa': 1.5, 'S': 1.0, 'K': 10.0,
    'mu': 0.4, 'memory_gain': 0.5, 'sigma': 0.5, 'delta_m': 0.05, 'eps_m': 0.01
}

x0_4d = np.array([0.1, 1.0, 0.5, 0.1])
res_eq = root(lambda x: np.array(system_vector_field_jax_4d(x, p_base)), x0_4d, method='hybr')
state_val = res_eq.x

J_state = np.array(jac_state_jax_4d(state_val, p_base))
tracker_init = HopfEigenTracker(np.ones(4) + 1j * np.ones(4))
target_idx, ev, _, right_vecs = tracker_init.track(J_state)

print(f"Equilibrium found. Initial target eigenvalue: {ev:.4f}")
```

```python
class HopfEigenTracker:
    def __init__(self, q_initial):
        self.q_current = q_initial / np.sqrt(np.real(np.vdot(q_initial, q_initial)))

    def track(self, J_state):
        evals, right_vecs = np.linalg.eig(J_state)
        q_prev_norm = self.q_current / np.sqrt(np.real(np.vdot(self.q_current, self.q_current)))

        scores = []
        for i in range(right_vecs.shape[1]):
            v = right_vecs[:, i]
            v_norm = v / np.sqrt(np.real(np.vdot(v, v)))
            phase_overlap = np.vdot(q_prev_norm, v_norm)
            v_corrected = v_norm * np.exp(-1j * np.angle(phase_overlap))
            scores.append(np.abs(np.vdot(q_prev_norm, v_corrected)))

        target_idx = np.argmax(scores)
        q_raw = right_vecs[:, target_idx]
        q_norm = q_raw / np.sqrt(np.real(np.vdot(q_raw, q_raw)))
        phase_overlap = np.vdot(q_prev_norm, q_norm)
        q_new = q_norm * np.exp(-1j * np.angle(phase_overlap))

        self.q_current = q_new / np.sqrt(np.real(np.vdot(q_new, q_new)))
        return target_idx, evals[target_idx], self.q_current, right_vecs

def bordered_hopf_augmented_system(z, p_base, tangent_n, z_predict, ds, q_ref, param_keys=('gamma', 'memory_gain')):
    x = z[:4]
    gamma_val, memory_gain_val, omega_val = z[4], z[5], z[6]
    qr, qi = z[7:11], z[11:15]
    q = qr + 1j * qi
    q_ref_norm = q_ref / np.sqrt(np.real(np.vdot(q_ref, q_ref)))

    p = p_base.copy()
    p[param_keys[0]] = gamma_val
    p[param_keys[1]] = memory_gain_val

    f_val = np.array(system_vector_field_jax_4d(x, p))
    J = np.array(jac_state_jax_4d(x, p))

    res_eigen_r = np.dot(J, qr) + omega_val * qi
    res_eigen_i = np.dot(J, qi) - omega_val * qr
    res_norm = np.real(np.vdot(q, q)) - 1.0
    res_phase = np.imag(np.vdot(q_ref_norm, q))
    res_arc = np.dot(tangent_n, z - z_predict)

    return np.hstack((f_val, res_eigen_r, res_eigen_i, [res_norm, res_phase, res_arc]))
```

```python
def compute_rigorous_kuznetsov_and_full_l2(state, omega_freq, J_state, right_vecs, target_idx, lambda_hopf, p_base, cur_gamma, cur_memory):
    if omega_freq < 1e-5:
        return 1e3, 0.0, False

    evals = np.linalg.eigvals(J_state)
    p_val = p_base.copy()
    p_val['gamma'] = cur_gamma
    p_val['memory_gain'] = cur_memory

    q = right_vecs[:, target_idx]
    q = q / np.sqrt(np.real(np.vdot(q, q)))

    evals_adj, left_vecs = np.linalg.eig(J_state.T.conj())
    adj_idx = np.argmin(np.abs(evals_adj - np.conj(lambda_hopf)))
    p = left_vecs[:, adj_idx]
    p = p / np.vdot(p, q)

    H = hess_state_jax_4d(state, p_val)
    T_tensor = third_state_jax_4d(state, p_val)

    B = lambda u, v: 0.5 * np.einsum('ijk,j,k->i', H, u, v)
    C = lambda u, v, w: (1.0 / 6.0) * np.einsum('ijkl,j,k,l->i', T_tensor, u, v, w)

    h11 = -np.linalg.solve(J_state, B(q, q.conj()))
    h20 = np.linalg.solve(2.0 * 1j * omega_freq * np.eye(4) - J_state, B(q, q))

    G21 = np.dot(p.conj(), C(q, q, q.conj()) + 2.0 * B(q.conj(), h20) + 4.0 * B(q, h11))
    l1 = np.real(G21) / (2.0 * omega_freq)

    try:
        matrix_3omega = 3.0 * 1j * omega_freq * np.eye(4) - J_state
        mat_0omega = -J_state
        matrix_2omega = 2.0 * 1j * omega_freq * np.eye(4) - J_state
        matrix_omega = 1j * omega_freq * np.eye(4) - J_state
        matrix_4omega = 4.0 * 1j * omega_freq * np.eye(4) - J_state

        h30 = np.linalg.solve(matrix_3omega, B(q, h20))
        h21 = np.linalg.solve(mat_0omega, 2.0 * B(q, h11) + B(q.conj(), h20))
        h22 = np.linalg.solve(matrix_2omega, B(q.conj(), h20) + 2.0 * B(q, h21))
        h31 = np.linalg.solve(matrix_omega, C(q, q, q.conj()) + 2.0 * B(q, h21) + B(q.conj(), h30) + B(h11, h20))
        h40 = np.linalg.solve(matrix_4omega, B(q, h30) + B(h20, h20))

        d_term_val = directional_d_tensor(state, p_val, q, q, q.conj(), q.conj())

        G32 = np.dot(
            p.conj(),
            d_term_val + 3.0 * C(q, q, h11) + 6.0 * C(q, q.conj(), h20) +
            2.0 * B(q.conj(), h30) + 3.0 * B(q, h21) + 3.0 * C(q, q.conj(), h21) +
            C(q.conj(), q.conj(), h30) + 3.0 * B(h20, h22) + 2.0 * B(h11, h31) + B(q, h40)
        )
        l2 = np.real(G32) / (12.0 * omega_freq)
    except Exception:
        l2 = 0.0

    return l1, l2, True
```

```python
# ==========================================
# CELL 6: ADJOINT TANGENT & ADAPTIVE ARC-LENGTH CONTINUATION
# ==========================================
jax_bordered_system_residual = jax.jit(lambda z, p_base, tangent_n, z_predict, ds, q_ref: jnp.concatenate([
    system_vector_field_jax_4d(z[:4], {**p_base, 'gamma': z[4], 'memory_gain': z[5]}),
    jnp.dot(jac_state_jax_4d(z[:4], {**p_base, 'gamma': z[4], 'memory_gain': z[5]}), z[7:11]) + z[6] * z[11:15],
    jnp.dot(jac_state_jax_4d(z[:4], {**p_base, 'gamma': z[4], 'memory_gain': z[5]}), z[11:15]) - z[6] * z[7:11],
    jnp.array([
        jnp.real(jnp.vdot(z[7:11] + 1j * z[11:15], z[7:11] + 1j * z[11:15])) - 1.0,
        jnp.imag(jnp.vdot(q_ref / jnp.sqrt(jnp.real(jnp.vdot(q_ref, q_ref))), z[7:11] + 1j * z[11:15])),
        jnp.dot(tangent_n, z - z_predict)
    ])
]))

jax_bordered_jacobian = jax.jit(jax.jacfwd(jax_bordered_system_residual, argnums=0))

def compute_ad_tangent(z, p_base, q_ref, tangent_old=None):
    tangent_dummy = np.zeros_like(z)
    J_mat = np.array(jax_bordered_jacobian(z, p_base, tangent_dummy, z, 0.0, jnp.array(q_ref)))
    _, s_vals, Vh = np.linalg.svd(J_mat)
    t = Vh[-1]
    t = t / np.linalg.norm(t)

    if tangent_old is not None and np.dot(t, tangent_old) < 0:
        t = -t
    return t, s_vals[-1]

def continue_hopf_curve_adaptive(z0, tangent0, p_base, max_steps=40, ds_init=0.01):
    branch = [z0]
    z = z0
    tangent = tangent0
    q_ref = z[7:11] + 1j * z[11:15]
    ds = ds_init

    k = 0
    while k < max_steps:
        omega_curr = z[6]
        if abs(omega_curr) < 1e-5:
            break

        z_predict = z + ds * tangent
        sol = root(
            lambda zz: bordered_hopf_augmented_system(zz, p_base, tangent, z_predict, ds, q_ref),
            z_predict,
            method='hybr'
        )

        if not sol.success or np.linalg.norm(sol.fun) > 1e-7:
            ds *= 0.5
            if ds < 1e-5:
                break
            continue

        z_new = sol.x
        q_new = z_new[7:11] + 1j * z_new[11:15]
        if np.real(np.vdot(q_ref, q_new)) < 0:
            z_new[7:11] *= -1.0
            z_new[11:15] *= -1.0
            q_new = -q_new

        q_ref = q_new
        tangent, sigma_min = compute_ad_tangent(z_new, p_base, q_ref, tangent_old=tangent)

        if sigma_min < 1e-6:
            ds = max(ds * 0.5, 1e-4)
        elif sigma_min > 0.05:
            ds = min(ds * 1.2, 0.05)

        branch.append(z_new)
        z = z_new
        k += 1
    return branch
```

```python
# ==========================================
# CELL 7: FULL 2x2 UNFOLDING MATRIX & BAUTIN OPTIMIZER
# ==========================================
def verify_rigorous_unfolding_matrix(z_bautin, p_base, q_ref):
    x = z_bautin[:4]
    gamma_0, mem_0, omega_0 = z_bautin[4], z_bautin[5], z_bautin[6]
    qr, qi = z_bautin[7:11], z_bautin[11:15]
    q = qr + 1j * qi
    q = q / np.sqrt(np.real(np.vdot(q, q)))

    p_v = p_base.copy()
    p_v['gamma'] = gamma_0
    p_v['memory_gain'] = mem_0

    J_loc = np.array(jac_state_jax_4d(x, p_v))
    evals, left_vecs = np.linalg.eig(J_loc.T.conj())
    evals_r, right_vecs = np.linalg.eig(J_loc)

    target_idx = np.argmin(np.abs(np.imag(evals_r) - omega_0))
    lambda_val = evals_r[target_idx]
    q_vec = right_vecs[:, target_idx]

    adj_idx = np.argmin(np.abs(evals - np.conj(lambda_val)))
    p_vec = left_vecs[:, adj_idx]
    p_vec = p_vec / np.vdot(p_vec, q_vec)

    f_extended = lambda st, p_dict: system_vector_field_jax_4d(st, p_dict)
    Fx = jax.jit(jax.jacfwd(lambda st: f_extended(st, p_v), argnums=0))(x)

    Fp = jax.jit(jax.jacfwd(lambda g, m: system_vector_field_jax_4d(x, {**p_base, 'gamma': g, 'memory_gain': m}), argnums=(0, 1)))(gamma_0, mem_0)
    Fp_mat = jnp.column_stack([Fp[0], Fp[1]])
    dx_dp = -np.linalg.solve(np.array(Fx), np.array(Fp_mat))

    J_p_jac = jax.jit(jax.jacfwd(lambda g, m: jac_state_jax_4d(x, {**p_base, 'gamma': g, 'memory_gain': m}), argnums=(0, 1)))(gamma_0, mem_0)
    Jx_dx_dp = np.dot(J_loc, dx_dp)

    d_lam_dp = np.array([
        np.dot(p_vec.conj(), np.dot(J_p_jac[0] + Jx_dx_dp[:, 0:1], q_vec)) / np.vdot(p_vec, q_vec),
        np.dot(p_vec.conj(), np.dot(J_p_jac[1] + Jx_dx_dp[:, 1:2], q_vec)) / np.vdot(p_vec, q_vec)
    ])

    eps_p = 1e-5
    def get_l1_at_params(g_val, m_val):
        p_temp = p_base.copy()
        p_temp['gamma'] = g_val
        p_temp['memory_gain'] = m_val
        st_eq = solve_equilibrium_4d(x, p_temp)
        J_ep = np.array(jac_state_jax_4d(st_eq, p_temp))
        tr_ep = HopfEigenTracker(q)
        t_idx, lam_ep, _, r_vecs_ep = tr_ep.track(J_ep)
        l1_ep, _, _ = compute_rigorous_kuznetsov_and_full_l2(st_eq, np.abs(np.imag(lam_ep)), J_ep, r_vecs_ep, t_idx, lam_ep, p_base, g_val, m_val)
        return l1_ep

    dl1_dgamma = (get_l1_at_params(gamma_0 + eps_p, mem_0) - get_l1_at_params(gamma_0 - eps_p, mem_0)) / (2.0 * eps_p)
    dl1_dmem = (get_l1_at_params(gamma_0, mem_0 + eps_p) - get_l1_at_params(gamma_0, mem_0 - eps_p)) / (2.0 * eps_p)

    _, l2_val, _ = compute_rigorous_kuznetsov_and_full_l2(x, omega_0, J_loc, right_vecs, target_idx, lambda_val, p_base, gamma_0, mem_0)

    unfolding_matrix = np.array([
        [np.real(d_lam_dp[0]), np.real(d_lam_dp[1])],
        [dl1_dgamma, dl1_dmem]
    ])
    return unfolding_matrix, np.linalg.det(unfolding_matrix), l2_val
```

```python
# ==========================================
# CELL 9: FULL AUTOMATED BAUTIN SEARCH & SWEEP LOOP
# ==========================================
def scan_and_locate_bautin(branch, p_base):
    bautin_results = []

    for i in range(len(branch) - 1):
        za, zb = branch[i], branch[i+1]

        def evaluate_l1_point_pure(zz):
            xp, gp, mp, om = zz[:4], zz[4], zz[5], zz[6]
            qp = zz[7:11] + 1j * zz[11:15]
            local_tr = HopfEigenTracker(qp)
            pp = p_base.copy()
            pp['gamma'] = gp
            pp['memory_gain'] = mp
            Jp = np.array(jac_state_jax_4d(xp, pp))
            t_idx, lam_val, _, _ = local_tr.track(Jp)
            l1_val, _, _ = compute_rigorous_kuznetsov_and_full_l2(xp, om, Jp, local_tr.q_current, t_idx, lam_val, p_base, gp, mp)
            return l1_val

        try:
            l1_a = evaluate_l1_point_pure(za)
            l1_b = evaluate_l1_point_pure(zb)
        except Exception:
            continue

        if l1_a * l1_b < 0:
            print(f"[BRACKET FOUND] Sign change detected between step {i} and {i+1} (l1: {l1_a:.4e} -> {l1_b:.4e})")
            z_guess_bautin = 0.5 * (za + zb)
            q_ref_bautin = z_guess_bautin[7:11] + 1j * z_guess_bautin[11:15]

            z_bautin, success = solve_bautin_point_pure(z_guess_bautin, p_base, q_ref_bautin, scale_l1_fixed=1e-2)
            if success:
                t_bautin, _ = compute_ad_tangent(z_bautin, p_base, z_bautin[7:11] + 1j * z_bautin[11:15])
                cross_val = hopf_crossing_derivative(z_bautin, t_bautin, p_base)
                unfolding_mat, unfolding_det, l2_val = verify_rigorous_unfolding_matrix(z_bautin, p_base, q_ref_bautin)

                if abs(cross_val) > 1e-6 and abs(unfolding_det) > 1e-6:
                    bautin_results.append({
                        'z_bautin': z_bautin,
                        'gamma': z_bautin[4],
                        'memory_gain': z_bautin[5],
                        'frequency': z_bautin[6],
                        'unfolding_matrix': unfolding_mat,
                        'unfolding_det': unfolding_det,
                        'l2': l2_val
                    })
                    print(f"[BAUTIN CONFIRMED] Gamma: {z_bautin[4]:.5f}, Memory Gain: {z_bautin[5]:.5f}, l2: {l2_val:.4e}")

    return bautin_results

print("Automated Bautin scan loop ready. Run: bautin_hits = scan_and_locate_bautin(branch, p_base)")
```

```python
# ==========================================
# CELL 10: 4D PHASE PORTRAIT & TIME-SERIES SIMULATION
# ==========================================
from scipy.integrate import solve_ivp

def simulate_system_dynamics(p_dict, initial_state, t_span=(0, 100), num_points=5000):
    t_eval = np.linspace(t_span[0], t_span[1], num_points)

    def rhs(t, state):
        return np.array(system_vector_field_jax_4d(state, p_dict))

    sol = solve_ivp(rhs, t_span, initial_state, t_eval=t_eval, method='Radau', rtol=1e-8, atol=1e-10)
    return sol.t, sol.y

def plot_phase_space_trajectories(sol_t, sol_y):
    fig = plt.figure(figsize=(14, 6))

    # A vs R Phase Plane
    ax1 = fig.add_subplot(121)
    ax1.plot(sol_y[0], sol_y[1], 'm-', lw=0.8)
    ax1.set_title("Phase Space: Activation (A) vs Resource (R)")
    ax1.set_xlabel("Activation ($A$)")
    ax1.set_ylabel("Resource ($R$)")
    ax1.grid(True)

    # Time Series of all 4 variables
    ax2 = fig.add_subplot(122)
    ax2.plot(sol_t, sol_y[0], label='A (Activation)', color='blue', lw=1)
    ax2.plot(sol_t, sol_y[1], label='R (Resource)', color='orange', lw=1)
    ax2.plot(sol_t, sol_y[2], label='N (Network/Node)', color='green', lw=1)
    ax2.plot(sol_t, sol_y[3], label='M (Memory)', color='purple', lw=1)
    ax2.set_title("System Dynamics Time Series")
    ax2.set_xlabel("Time ($t$)")
    ax2.set_ylabel("State Values")
    ax2.legend(loc='upper right')
    ax2.grid(True)

    plt.tight_layout()
    plt.show()

print("Simulation module ready. Run: t, y = simulate_system_dynamics(p_base, state_val); plot_phase_space_trajectories(t, y)")
```

```python
# ==========================================
# CELL 11: BIFURCATION DIAGRAM & PHASE PORTRAIT GENERATOR
# ==========================================
def generate_bifurcation_summary_plots(branch, bautin_hits, p_base):
    gamma_vals = [z[4] for z in branch]
    memory_vals = [z[5] for z in branch]
    omega_vals = [z[6] for z in branch]

    fig = plt.figure(figsize=(16, 6))

    # Plot 1: 3D Parameter-Frequency Space
    ax1 = fig.add_subplot(121, projection='3d')
    ax1.plot(gamma_vals, memory_vals, omega_vals, label='Hopf Branch', color='blue', lw=2)

    if bautin_hits:
        b_gammas = [b['gamma'] for b in bautin_hits]
        b_mems = [b['memory_gain'] for b in bautin_hits]
        b_freqs = [b['frequency'] for b in bautin_hits]
        ax1.scatter(b_gammas, b_mems, b_freqs, color='red', s=100, label='Bautin (Generalized Hopf)', depthshade=False)

    ax1.set_title("3D Hopf Curve & Bautin Points")
    ax1.set_xlabel("Gamma ($\gamma$)")
    ax1.set_ylabel("Memory Gain ($m$)")
    ax1.set_zlabel("Frequency ($\omega$)")
    ax1.legend()

    # Plot 2: Parameter Space Projection with Bautin Markers
    ax2 = fig.add_subplot(122)
    ax2.plot(gamma_vals, memory_vals, 'k--', label='Hopf Bifurcation Locus', lw=1.5)

    if bautin_hits:
        for b in bautin_hits:
            ax2.plot(b['gamma'], b['memory_gain'], 'r*', markersize=14, label='Bautin Point')
            ax2.annotate(f"l2: {b['l2']:.2e}\nDet: {b['unfolding_det']:.2e}",
                         (b['gamma'], b['memory_gain']),
                         textcoords="offset points", xytext=(0, 10), ha='center',
                         bbox=dict(boxstyle="round,pad=0.3", fc="yellow", alpha=0.5))

    ax2.set_title("2D Parameter Projection & Unfolding Validation")
    ax2.set_xlabel("Gamma ($\gamma$)")
    ax2.set_ylabel("Memory Gain ($m$)")
    ax2.grid(True)
    ax2.legend()

    plt.tight_layout()
    plt.show()

print("Bifurcation summary visualization module ready. Run: generate_bifurcation_summary_plots(branch, bautin_hits, p_base)")
```

```python
# ==========================================
# CELL 12: EXPORTING CONTINUATION & BAUTIN DATA
# ==========================================
import json

def export_bautin_results_to_json(branch, bautin_hits, filename="bautin_pipeline_results.json"):
    export_data = {
        "branch_length": len(branch),
        "branch_points": [
            {
                "step": i,
                "state": z[:4].tolist(),
                "gamma": float(z[4]),
                "memory_gain": float(z[5]),
                "omega": float(z[6])
            } for i, z in enumerate(branch)
        ],
        "bautin_points": [
            {
                "gamma": float(b['gamma']),
                "memory_gain": float(b['memory_gain']),
                "frequency": float(b['frequency']),
                "l2": float(b['l2']),
                "unfolding_determinant": float(b['unfolding_det']),
                "unfolding_matrix": b['unfolding_matrix'].tolist()
            } for b in bautin_hits
        ]
    }

    with open(filename, 'w') as f:
        json.dump(export_data, f, indent=4)
    print(f"Pipeline results successfully exported to {filename}")

print("Export utility ready. Run: export_bautin_results_to_json(branch, bautin_hits)")
```

```python
# ==========================================
# CELL 13: EIGENVALUE SPECTRUM SWEEP & STABILITY ANALYSIS
# ==========================================
def analyze_parameter_spectrum_sweep(p_base, gamma_range, memory_val):
    gamma_vals = np.linspace(gamma_range[0], gamma_range[1], 50)
    max_real_evals = []
    frequencies = []

    x_guess = np.array([0.1, 1.0, 0.5, 0.1])

    for g in gamma_vals:
        p_test = p_base.copy()
        p_test['gamma'] = g
        p_test['memory_gain'] = memory_val

        try:
            st = solve_equilibrium_4d(x_guess, p_test)
            x_guess = st
            J = np.array(jac_state_jax_4d(st, p_test))
            evals = np.linalg.eigvals(J)

            max_real_evals.append(np.max(np.real(evals)))
            complex_evals = [ev for ev in evals if abs(np.imag(ev)) > 1e-4]
            frequencies.append(np.abs(complex_evals[0]) if complex_evals else 0.0)
        except Exception:
            max_real_evals.append(np.nan)
            frequencies.append(np.nan)

    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    axes[0].plot(gamma_vals, max_real_evals, 'b-', lw=1.5)
    axes[0].axhline(0, color='r', linestyle='--', alpha=0.7, label='Stability Boundary')
    axes[0].set_title("Max Real Eigenvalue vs. Gamma ($\gamma$)")
    axes[0].set_xlabel("Gamma ($\gamma$)")
    axes[0].set_ylabel("Max Re($\lambda$)")
    axes[0].grid(True)
    axes[0].legend()

    axes[1].plot(gamma_vals, frequencies, 'g-', lw=1.5)
    axes[1].set_title("Imaginary Frequency vs. Gamma ($\gamma$)")
    axes[1].set_xlabel("Gamma ($\gamma$)")
    axes[1].set_ylabel("Frequency ($\omega$)")
    axes[1].grid(True)

    plt.tight_layout()
    plt.show()

print("Spectrum sweep module ready. Run: analyze_parameter_spectrum_sweep(p_base, (0.5, 3.0), 0.5)")
```

```python
# ==========================================
# CELL 14: MASTER PIPELINE ORCHESTRATOR
# ==========================================
def run_master_bautin_pipeline():
    print("=" * 60)
    print("STARTING MASTER BAUTIN COMPUTATIONAL PIPELINE")
    print("=" * 60)

    # 1. Base Parameters & Equilibrium
    p_base = {
        'alpha': 1.2, 'beta': 0.8, 'gamma': 1.0, 'rho': 0.6, 'R_max': 5.0,
        'eta': 0.5, 'omega': 0.1, 'kappa': 1.5, 'S': 1.0, 'K': 10.0,
        'mu': 0.4, 'memory_gain': 0.5, 'sigma': 0.5, 'delta_m': 0.05, 'eps_m': 0.01
    }

    x0_4d = np.array([0.1, 1.0, 0.5, 0.1])
    res_eq = root(lambda x: np.array(system_vector_field_jax_4d(x, p_base)), x0_4d, method='hybr')
    state_val = res_eq.x

    # 2. Initial Hopf Tracking
    J_state = np.array(jac_state_jax_4d(state_val, p_base))
    tracker_init = HopfEigenTracker(np.ones(4) + 1j * np.ones(4))
    target_idx, ev, q_init, right_vecs = tracker_init.track(J_state)

    # Construct initial extended state vector z0 for continuation: [x(4), gamma, memory_gain, omega, qr(4), qi(4)]
    z0 = np.hstack((
        state_val,
        np.array([p_base['gamma'], p_base['memory_gain'], np.abs(np.imag(ev))]),
        np.real(q_init),
        np.imag(q_init)
    ))

    # 3. Compute Initial Tangent & Run Branch Continuation
    tangent0, _ = compute_ad_tangent(z0, p_base, q_init)
    print("Running adaptive Hopf curve continuation...")
    branch = continue_hopf_curve_adaptive(z0, tangent0, p_base, max_steps=35, ds_init=0.01)
    print(f"Continuation complete. Total branch points generated: {len(branch)}")

    # 4. Scan and Locate Bautin Points
    print("Scanning branch for Bautin (generalized Hopf) bifurcation points...")
    bautin_hits = scan_and_locate_bautin(branch, p_base)
    print(f"Bautin scan complete. Total generalized Hopf points isolated: {len(bautin_hits)}")

    # 5. Export and Summarize
    if bautin_hits:
        export_bautin_results_to_json(branch, bautin_hits)
        generate_bifurcation_summary_plots(branch, bautin_hits, p_base)
    else:
        print("No Bautin sign changes crossed in this segment range. Try adjusting baseline parameter intervals.")

    print("=" * 60)
    print("PIPELINE EXECUTION FINISHED SUCCESSFULLY")
    print("=" * 60)
    return branch, bautin_hits

print("Master orchestrator ready. Run: branch, bautin_hits = run_master_bautin_pipeline()")
```
