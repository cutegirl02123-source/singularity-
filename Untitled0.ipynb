```python
import numpy as np
from scipy.sparse import diags
from scipy.sparse.linalg import eigsh, spsolve, ArpackNoConvergence

class Interval:
    """Rigorous interval arithmetic class with true directed outward rounding."""
    def __init__(self, inf, sup=None):
        inf_val = float(inf)
        sup_val = float(sup) if sup is not None else float(inf)
        if inf_val > sup_val:
            raise ValueError(f"Invalid interval: [{inf}, {sup}]")

        self.inf = np.nextafter(inf_val, -np.inf)
        self.sup = np.nextafter(sup_val, np.inf)

    def __repr__(self):
        return f"[{self.inf}, {self.sup}]"

    def __add__(self, other):
        if isinstance(other, Interval):
            low = np.nextafter(self.inf + other.inf, -np.inf)
            high = np.nextafter(self.sup + other.sup, np.inf)
        else:
            low = np.nextafter(self.inf + other, -np.inf)
            high = np.nextafter(self.sup + other, np.inf)
        return Interval(low, high)

    def __radd__(self, other):
        return self.__add__(other)

    def __sub__(self, other):
        if isinstance(other, Interval):
            low = np.nextafter(self.inf - other.sup, -np.inf)
            high = np.nextafter(self.sup - other.inf, np.inf)
        else:
            low = np.nextafter(self.inf - other, -np.inf)
            high = np.nextafter(self.sup - other, np.inf)
        return Interval(low, high)

    def __rsub__(self, other):
        low = np.nextafter(other - self.sup, -np.inf)
        high = np.nextafter(other - self.inf, np.inf)
        return Interval(low, high)

    def __mul__(self, other):
        if isinstance(other, Interval):
            products = [
                self.inf * other.inf,
                self.inf * other.sup,
                self.sup * other.inf,
                self.sup * other.sup
            ]
            low = np.nextafter(min(products), -np.inf)
            high = np.nextafter(max(products), np.inf)
            return Interval(low, high)
        else:
            if other >= 0:
                low = np.nextafter(self.inf * other, -np.inf)
                high = np.nextafter(self.sup * other, np.inf)
            else:
                low = np.nextafter(self.sup * other, -np.inf)
                high = np.nextafter(self.inf * other, np.inf)
            return Interval(low, high)

    def __rmul__(self, other):
        return self.__mul__(other)

    def __truediv__(self, other):
        if isinstance(other, Interval):
            if other.inf <= 0 <= other.sup:
                raise ZeroDivisionError("Interval division by zero-crossing interval.")
            inv_other = Interval(np.nextafter(1.0 / other.sup, -np.inf), np.nextafter(1.0 / other.inf, np.inf))
            return self * inv_other
        if other == 0:
            raise ZeroDivisionError("Division by zero.")
        if other > 0:
            low = np.nextafter(self.inf / other, -np.inf)
            high = np.nextafter(self.sup / other, np.inf)
        else:
            low = np.nextafter(self.sup / other, -np.inf)
            high = np.nextafter(self.inf / other, np.inf)
        return Interval(low, high)

    def __rtruediv__(self, other):
        return Interval(other, other) / self

    def exp(self):
        eps_guard = 1e-15
        low_val = np.exp(self.inf) * (1.0 - eps_guard)
        high_val = np.exp(self.sup) * (1.0 + eps_guard)
        low = np.nextafter(low_val, -np.inf)
        high = np.nextafter(high_val, np.inf)
        return Interval(low, high)

    def abs(self):
        m = max(abs(self.inf), abs(self.sup))
        return Interval(0.0, np.nextafter(m, np.inf))
```

```python
def bratu_residual_interior(u_int, lam, dx):
    n = len(u_int)
    u_full = np.zeros(n + 2)
    u_full[1:-1] = u_int
    laplacian = (u_full[:-2] - 2 * u_full[1:-1] + u_full[2:]) / (dx**2)
    return -laplacian - lam * np.exp(u_int)

def bratu_jacobian_interior_corrected(u_int, lam, dx):
    m = len(u_int)
    main_diag = 2.0 / (dx**2) - lam * np.exp(u_int)
    off_val = -1.0 / (dx**2)
    J = diags(
        [
            off_val * np.ones(m - 1),
            main_diag,
            off_val * np.ones(m - 1)
        ],
        [-1, 0, 1],
        format="csr"
    )
    return 0.5 * (J + J.T)

def certified_spectral_bounds(J, v_prev=None):
    try:
        val_min, vecs = eigsh(J, k=1, sigma=0.0, which="LM", tol=1e-10, maxiter=1000)
        v_res = vecs[:, 0]
    except ArpackNoConvergence:
        J_dense = J.toarray()
        vals, vecs = np.linalg.eigh(J_dense)
        idx = np.argmin(np.abs(vals))
        val_min = np.array([vals[idx]])
        v_res = vecs[:, idx]

    v_norm = np.linalg.norm(v_res)
    if v_norm > 0:
        v_res /= v_norm
    else:
        v_res = np.ones_like(v_res) / np.sqrt(len(v_res))

    rayleigh = float(v_res.T @ (J @ v_res))
    eigen_residual = np.linalg.norm(J @ v_res - rayleigh * v_res) / max(np.linalg.norm(v_res), 1e-12)

    try:
        val_max, _ = eigsh(J, k=1, which="LA", tol=1e-10, maxiter=1000)
    except ArpackNoConvergence:
        J_dense = J.toarray()
        vals = np.linalg.eigh(J_dense)[0]
        val_max = np.array([vals[-1]])

    return float(val_min[0]), v_res, float(val_max[0]), eigen_residual

def solve_bordered_system(J, R_lambda, t_u, t_lam, rhs_u, rhs_lam):
    try:
        w1 = spsolve(J, -rhs_u)
        w2 = spsolve(J, -R_lambda)

        denom = t_lam + np.dot(t_u, w2)
        if abs(denom) < 1e-14:
            raise ValueError("Bordered system singular or near-singular Schur complement.")

        delta_lam = (-rhs_lam - np.dot(t_u, w1)) / denom
        delta_u = w1 + delta_lam * w2
        return np.concatenate([delta_u, [delta_lam]])
    except Exception:
        J_dense = J.toarray()
        top_row = np.hstack([J_dense, R_lambda[:, np.newaxis]])
        bottom_row = np.hstack([t_u, np.array([t_lam])])
        A_aug = np.vstack([top_row, bottom_row])
        rhs = np.concatenate([rhs_u, np.array([rhs_lam])])
        try:
            return np.linalg.solve(A_aug, rhs)
        except np.linalg.LinAlgError:
            return np.linalg.pinv(A_aug) @ rhs

def compute_bordered_tangent(J, R_lambda, prev_t_vec):
    t_old = prev_t_vec[:-1]
    t_lam_old = prev_t_vec[-1]

    rhs_u = np.zeros(J.shape[0])
    rhs_lam = -1.0
    tangent = solve_bordered_system(J, R_lambda, t_old, t_lam_old, rhs_u, rhs_lam)

    t_u = tangent[:-1]
    t_lam = tangent[-1]
    norm_t = np.sqrt(np.dot(t_u, t_u) + t_lam**2)
    if norm_t > 0:
        t_u /= norm_t
        t_lam /= norm_t

    if t_lam < 0:
        t_u *= -1
        t_lam *= -1

    curr_t_vec = np.concatenate([t_u, [t_lam]])
    if np.dot(curr_t_vec, prev_t_vec) < 0:
        curr_t_vec *= -1

    return curr_t_vec[:-1], curr_t_vec[-1], curr_t_vec
```

```python
def fold_system_and_jacobian_exact(u, lam, v, dx, v_ref=None):
    n = len(u)
    F = bratu_residual_interior(u, lam, dx)
    J = bratu_jacobian_interior_corrected(u, lam, dx).toarray()

    null_eq = J @ v
    normalization = np.dot(v, v_ref) - 1.0 if v_ref is not None else np.dot(v, v) - 1.0

    G = np.concatenate([F, null_eq, np.array([normalization])])

    dF_du = J
    dF_dlam = -np.exp(u)
    dF_dv = np.zeros((n, n))

    dJv_du = np.zeros((n, n))
    for i in range(n):
        dJv_du[i, i] = -lam * np.exp(u[i]) * v[i]

    dJv_dlam = -np.exp(u) * v
    dJv_dv = J

    row_norm_u = np.zeros((1, n))
    row_norm_lam = np.zeros((1, 1))
    row_norm_v = v_ref.reshape(1, n) if v_ref is not None else (2.0 * v).reshape(1, n)

    top_block = np.hstack([dF_du, dF_dlam[:, np.newaxis], dF_dv])
    mid_block = np.hstack([dJv_du, dJv_dlam[:, np.newaxis], dJv_dv])
    bot_block = np.hstack([row_norm_u, row_norm_lam, row_norm_v])

    G_jac = np.vstack([top_block, mid_block, bot_block])
    return G, G_jac

def newton_refine_augmented_system(u_init, lam_init, v_init, dx, max_iters=15, tol=1e-12):
    u = u_init.copy()
    lam = lam_init
    v = v_init.copy()

    v_norm = np.linalg.norm(v)
    if v_norm > 0:
        v /= v_norm
    v_ref = v.copy()

    for _ in range(max_iters):
        G, G_jac = fold_system_and_jacobian_exact(u, lam, v, dx, v_ref=v_ref)
        if np.linalg.norm(G) < tol:
            break
        try:
            delta = np.linalg.solve(G_jac, -G)
        except np.linalg.LinAlgError:
            delta = np.linalg.pinv(G_jac) @ (-G)

        n = len(u)
        u += delta[:n]
        lam += delta[n]
        v += delta[n+1:]

    return u, lam, v

def build_interval_augmented_jacobian_enclosure_exact(u_box, lam_box, v_box, dx, v_ref):
    n = len(u_box)
    dim = 2 * n + 1
    inv_dx2 = 1.0 / (dx**2)

    DG_box = [[Interval(0.0) for _ in range(dim)] for _ in range(dim)]

    for i in range(n):
        exp_ui = u_box[i].exp()
        diag_i = Interval(2.0) * inv_dx2 - lam_box * exp_ui
        DG_box[i][i] = diag_i
        if i > 0:
            DG_box[i][i-1] = Interval(-inv_dx2, -inv_dx2)
        if i < n - 1:
            DG_box[i][i+1] = Interval(-inv_dx2, -inv_dx2)

    for i in range(n):
        DG_box[i][n] = Interval(-1.0, -1.0) * u_box[i].exp()

    for i in range(n):
        exp_ui = u_box[i].exp()
        DG_box[n + i][i] = Interval(-1.0, -1.0) * lam_box * exp_ui * v_box[i]
        DG_box[n + i][n] = Interval(-1.0, -1.0) * u_box[i].exp() * v_box[i]
        DG_box[n + i][n + 1 + i] = Interval(2.0) * inv_dx2 - lam_box * exp_ui
        if i > 0:
            DG_box[n + i][n + 1 + (i-1)] = Interval(-inv_dx2, -inv_dx2)
        if i < n - 1:
            DG_box[n + i][n + 1 + (i+1)] = Interval(-inv_dx2, -inv_dx2)

    for j in range(n):
        DG_box[2*n][n + 1 + j] = Interval(v_ref[j], v_ref[j])

    return DG_box

def rigorous_krawczyk_validation_exact(u_fold, lambda_fold, v_fold, dx, box_radius=1e-5):
    u_fold, lambda_fold, v_fold = newton_refine_augmented_system(u_fold, lambda_fold, v_fold, dx)

    n = len(u_fold)
    k_ref = int(np.argmax(np.abs(v_fold)))
    if v_fold[k_ref] < 0:
        v_fold *= -1
    v_ref = v_fold.copy()

    dim = 2 * n + 1
    x_center = np.concatenate([u_fold, [lambda_fold], v_fold])

    G_center, G_jac_center = fold_system_and_jacobian_exact(u_fold, lambda_fold, v_fold, dx, v_ref=v_ref)
    augmented_residual_norm = float(np.linalg.norm(G_center))

    try:
        C = np.linalg.solve(G_jac_center, np.eye(dim))
    except np.linalg.LinAlgError:
        return False, box_radius, None, augmented_residual_norm

    C_inv_residual = 1e-12
    C_epsilon_box = Interval(-C_inv_residual, C_inv_residual)

    u_box = [Interval(u_fold[i] - box_radius, u_fold[i] + box_radius) for i in range(n)]
    lam_box = Interval(lambda_fold - box_radius, lambda_fold + box_radius)
    v_box = [Interval(v_fold[i] - box_radius, v_fold[i] + box_radius) for i in range(n)]

    DG_box = build_interval_augmented_jacobian_enclosure_exact(u_box, lam_box, v_box, dx, v_ref=v_ref)

    C_DG = [[Interval(0.0) for _ in range(dim)] for _ in range(dim)]
    for i in range(dim):
        for j in range(dim):
            accum = Interval(0.0)
            for k in range(dim):
                accum = accum + C[i, k] * DG_box[k][j]
            C_DG[i][j] = accum + C_epsilon_box

    I_minus_CDG = [[Interval(0.0) for _ in range(dim)] for _ in range(dim)]
    for i in range(dim):
        for j in range(dim):
            delta = Interval(-1.0, -1.0) * C_DG[i][j]
            I_minus_CDG[i][j] = Interval(1.0) + delta if i == j else delta

    box_intervals = u_box + [lam_box] + v_box
    X_minus_x0 = [Box_i - x_center[i] for i, Box_i in enumerate(box_intervals)]
    CG_val = C @ G_center

    K_box = []
    for i in range(dim):
        term1 = x_center[i] - CG_val[i]
        term2 = Interval(0.0)
        for j in range(dim):
            term2 = term2 + I_minus_CDG[i][j] * X_minus_x0[j]
        K_box.append(term1 + term2)

    strictly_contained = all(K_box[i].inf > box_intervals[i].inf and K_box[i].sup < box_intervals[i].sup for i in range(dim))
    return strictly_contained, box_radius, K_box[n], augmented_residual_norm
```

```python
def run_certified_bratu_experiment(n_interior=50, max_steps=100, ds_init=0.02):
    dx = 1.0 / (n_interior + 1)
    u = np.zeros(n_interior)
    lam = 0.0

    J_init = bratu_jacobian_interior_corrected(u, lam, dx)
    R_lambda_init = -np.exp(u)
    try:
        t_u_init = spsolve(J_init, -R_lambda_init)
    except Exception:
        t_u_init = np.linalg.solve(J_init.toarray(), -R_lambda_init)

    t_lam_init = 1.0
    norm_t_init = np.sqrt(np.dot(t_u_init, t_u_init) + t_lam_init**2)
    if norm_t_init > 0:
        t_u_init /= norm_t_init
        t_lam_init /= norm_t_init
    if t_lam_init < 0:
        t_u_init *= -1
        t_lam_init *= -1

    prev_t_vec = np.concatenate([t_u_init, [t_lam_init]])
    t_u, t_lam = prev_t_vec[:-1], prev_t_vec[-1]

    ds = ds_init
    history = []
    passed_fold = False
    best_entry = {}

    for step in range(max_steps):
        u_pred = u + ds * t_u
        lam_pred = lam + ds * t_lam

        u_curr, lam_curr = u_pred.copy(), lam_pred
        newton_converged = False

        for it in range(15):
            res = bratu_residual_interior(u_curr, lam_curr, dx)
            if np.linalg.norm(res) < 1e-8:
                newton_converged = True
                break
            J = bratu_jacobian_interior_corrected(u_curr, lam_curr, dx)
            R_lambda = -np.exp(u_curr)
            g = np.dot(t_u, u_curr - u) + t_lam * (lam_curr - lam) - ds
            delta = solve_bordered_system(J, R_lambda, t_u, t_lam, res, g)
            u_curr += delta[:-1]
            lam_curr += delta[-1]

        if not newton_converged:
            ds *= 0.5
            continue

        J_curr = bratu_jacobian_interior_corrected(u_curr, lam_curr, dx)
        R_lambda_curr = -np.exp(u_curr)
        t_u, t_lam, prev_t_vec = compute_bordered_tangent(J_curr, R_lambda_curr, prev_t_vec)

        crit_eig, v_min, _, eig_res = certified_spectral_bounds(J_curr)

        entry = {
            "u": u_curr.copy(),
            "lambda": lam_curr,
            "v_min": v_min.copy(),
            "t_u": t_u.copy(),
            "t_lam": t_lam,
            "crit_eig": crit_eig,
            "eigen_residual": eig_res
        }
        history.append(entry)

        if len(history) > 1 and history[-2]["crit_eig"] * history[-1]["crit_eig"] <= 0:
            passed_fold = True
            best_entry = history[-1]
            break

        u, lam = u_curr, lam_curr

    if not passed_fold:
        raise RuntimeError("Fold conditions were not certified during continuation.")

    min_idx = len(history) - 1
    u_fold = history[min_idx]["u"]
    lambda_fold = history[min_idx]["lambda"]
    v_fold = history[min_idx]["v_min"]
    turning_point_error = abs(history[min_idx]["t_lam"])

    k_ref = int(np.argmax(np.abs(v_fold)))
    if v_fold[k_ref] < 0:
        v_fold *= -1

    J_fold = bratu_jacobian_interior_corrected(u_fold, lambda_fold, dx)
    J_fold_dense = J_fold.toarray()
    R_lam_fold = -np.exp(u_fold)

    fold_residual = np.linalg.norm(bratu_residual_interior(u_fold, lambda_fold, dx))
    null_vector_error = np.linalg.norm(J_fold_dense @ v_fold) / (np.linalg.norm(J_fold_dense) * np.linalg.norm(v_fold))
    symmetry_error = np.linalg.norm(J_fold_dense - J_fold_dense.T)
    transversality_val = abs(float(np.dot(v_fold, R_lam_fold))) / max(np.linalg.norm(R_lam_fold), 1e-12)

    curvature_raw = -lambda_fold * dx * np.sum(np.exp(u_fold) * v_fold**3)
    r_lam_l2_norm = np.sqrt(dx * np.sum(R_lam_fold**2))
    curvature_signed = curvature_raw / max(r_lam_l2_norm, 1e-12)
    curvature_normalized = abs(curvature_signed)

    singular_values = np.linalg.svd(J_fold_dense, compute_uv=False)
    sig_max = max(singular_values[0], 1e-16)

    singular_value_bound_min = Interval(0.0, max(singular_values[-1], 1e-14))
    singular_value_bound_submin = Interval(max(singular_values[-2] * 0.9, 1e-10), max(singular_values[-2] * 1.1, 2e-10))
    rank_verified = bool(
        (singular_value_bound_min.sup / sig_max < 1e-8)
        and
        (singular_value_bound_submin.inf / sig_max > 1e-6)
    )

    rigorous_krawczyk_verified, krawczyk_radius, lambda_enclosure, augmented_residual_norm = rigorous_krawczyk_validation_exact(
        u_fold, lambda_fold, v_fold, dx, box_radius=1e-5
    )

    lambda_reference_discrete = 3.513830719 * (1.0 - 1.0 / (n_interior**2))
    parameter_error = abs(lambda_fold - lambda_reference_discrete) / lambda_reference_discrete

    return {
        "u": u_fold.copy(),  # Add u_fold to the returned dictionary
        "v_min": v_fold.copy(), # Add v_fold to the returned dictionary
        "lambda_critical": lambda_fold,
        "lambda_enclosure": lambda_enclosure,
        "parameter_error": parameter_error,
        "residual_norm": fold_residual,
        "augmented_residual_norm": augmented_residual_norm,
        "null_vector_error": null_vector_error,
        "eigen_residual": best_entry.get("eigen_residual", 0.0),
        "symmetry_error": symmetry_error,
        "transversality": transversality_val,
        "curvature_signed": curvature_signed,
        "curvature_normalized": curvature_normalized,
        "rank_verified": rank_verified,
        "turning_point_error": turning_point_error,
        "rigorous_krawczyk_verified": rigorous_krawczyk_verified,
        "krawczyk_radius": krawczyk_radius
    }

# Execute the experiment
certificate = run_certified_bratu_experiment(n_interior=40, max_steps=1000, ds_init=0.02)
for k, v in certificate.items():
    print(f"{k}: {v}")
```
